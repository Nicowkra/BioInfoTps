### f) ¿Con qué residuos interaccionan ahora las cadenas laterales? ¿Qué tipo de aminoácidos son estos? ¿Qué tipo de interacción establecen?

Al medir el entorno de la Fosfotreonina 183 y la Fosfotirosina 185 en la ERK2 activa, se observa un cambio en sus interacciones:

* **Residuos interactuantes:** Las cadenas laterales fosforiladas ahora interactúan con nuevos residuos vecinos, destacándose fuertemente la presencia de Argininas y Lisinas.
* **Tipo de aminoácidos:** Las Argininas y Lisinas son aminoácidos básicos, cuyas cadenas laterales poseen carga positiva.
* **Tipo de interacción:** Debido a que los grupos fosfato introducen cargas muy negativas y los aminoácidos circundantes tienen cargas positivas, la interacción principal que se establece es electrostática, además de puentes de hidrógeno.

### g) Finalmente, reflexione sobre el mecanismo molecular de activación de ERK2 y proporcione una breve explicación

Comparando las dos estructuras, podemos entender cómo se activa ERK2.  
En la forma inactiva, los residuos Thr183 y Tyr185 del loop de activación apuntan hacia adentro y, según las distancias que medí, quedan sujetos por interacciones polares con residuos vecinos. Eso mantiene el loop en una posición que impide el acceso al sitio donde se une el ATP.  
Cuando la proteína se fosforila, la Thr183 y la Tyr185 pasan a ser fosfotreonina y fosfotirosina, incorporando grupos fosfato con mucha carga negativa. Al medir el entorno en la forma activa vi que esos fosfatos ya no interactúan con los mismos residuos, sino que quedan cerca de argininas y lisinas (aminoácidos básicos, con carga positiva). Esa atracción electrostática reordena el loop de activación y lo corre de lugar.  
El resultado es que el loop deja de tapar el sitio catalítico y la proteína queda en una conformación capaz de unir ATP y fosforilar a sus sustratos.
