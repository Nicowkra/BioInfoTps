8) __Vaya a “Navigate”, seleccione “Organisms” y luego elija el organismo Homo sapiens. Observe la cantidad de genes en subsistemas ¿Es igual a lo observado para Exiguobacterium sp. S17? ¿Por qué cree que pueden darse estas diferencias?__  
![Homo Sapiens](Img/Ej2_Homo.png)
Definitivamente no. Al observar el genoma de Homo sapiens, notamos que de las 33869 secuencias codificantes detectadas, solo el 5% logró ser asignado a un subsistema. Esto contrasta drásticamente con Exiguobacterium sp. S17, donde se logró asignar un 40% de sus genes a subsistemas.  
Estas diferencias se deben al diseño de RAST, que es una plataforma de anotación automática de genomas procariotas. Su pipeline funciona comparando las secuencias contra una librería curada manualmente de familias de proteínas. Dado que Homo sapiens es un organismo eucariota, evolutivamente está extremadamente alejado de las bacterias, por lo que a mayor divergencia, la cobertura es menor.
